```python
# Program to perform arithmetic and statistical calculations
import math
```

```python

# Arithmetic operations
a = 20
b = 10

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus (Remainder):", a % b)
print("Exponentiation (Power):", a ** b)
```

```python

print("\n----- Mathematical Operations -----")

print("Square Root:", math.sqrt(a))
print("Square:", a ** 2)
print("Cube:", a ** 3)
print("Absolute Value:", abs(-20))
print("Factorial:", math.factorial(5))
print("Power:", math.pow(a, 2))
print("Maximum:", max(a, b))
print("Minimum:", min(a, b))
print("Round:", round(10.567, 2))
```

```python
print("\n----- Statistical Operations -----")

numbers = [10, 20, 30, 40, 50]

print("Sum:", sum(numbers))
print("Mean:", sum(numbers) / len(numbers))
print("Median:", sorted(numbers)[len(numbers) // 2])
print("Maximum:", max(numbers))
print("Minimum:", min(numbers))
print("Count:", len(numbers))
```

```python
from statistics import mean, median, stdev

numbers = [10, 20, 30, 40, 50]

print("Mean:", mean(numbers))
print("Median:", median(numbers))
print("Standard Deviation:", stdev(numbers))
```

```python
print("\n----- NumPy Statistical Operations -----")

import numpy as np

print("Mean:", np.mean(numbers))
print("Median:", np.median(numbers))
print("Standard Deviation:", np.std(numbers))
print("Variance:", np.var(numbers))
print("Sum:", np.sum(numbers))
print("Maximum:", np.max(numbers))
print("Minimum:", np.min(numbers))
print("25th Percentile:", np.percentile(numbers, 25))
print("75th Percentile:", np.percentile(numbers, 75))
```
