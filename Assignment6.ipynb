```python
import pandas as pd
import numpy as np

df = pd.read_csv("iris.csv")

print(df.head())
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDataset Information:")
print(df.info())
```

```python
# Create a copy of the dataset
df1 = df.copy()

# Add missing values for demonstration
df1.loc[0, 'sepal_length'] = np.nan
df1.loc[5, 'sepal_width'] = np.nan

print("Missing Values Before Handling:")
print(df1.isnull().sum())

# Fill missing values with column mean
df1['sepal_length'] = df1['sepal_length'].fillna(df1['sepal_length'].mean())
df1['sepal_width'] = df1['sepal_width'].fillna(df1['sepal_width'].mean())

print("\nMissing Values After Handling:")
print(df1.isnull().sum())
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df1['species'] = encoder.fit_transform(df1['species'])

print("Dataset After Encoding:")
print(df1.head())

print("\nSpecies Mapping:")
for i, name in enumerate(encoder.classes_):
    print(name, "=", i)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

numeric_columns = [
    'sepal_length',
    'sepal_width',
    'petal_length',
    'petal_width'
]

df1[numeric_columns] = scaler.fit_transform(df1[numeric_columns])

print("Dataset After Normalization:")
print(df1.head())
```

```python
print("Final Preprocessed Dataset:")
print(df1.head())

print("\nDataset Shape:")
print(df1.shape)

print("\nMissing Values:")
print(df1.isnull().sum())
```
