```python
import math

a = 25
b = 5

print("Square Root:", math.sqrt(a))
print("Power:", math.pow(a, 2))
print("Factorial:", math.factorial(b))
print("Absolute Value:", math.fabs(-25))
print("Ceiling:", math.ceil(4.3))
print("Floor:", math.floor(4.8))
print("Logarithm:", math.log(10))
print("Value of Pi:", math.pi)
```

```python
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

print("Array:", numbers)
print("Sum:", np.sum(numbers))
print("Mean:", np.mean(numbers))
print("Median:", np.median(numbers))
print("Standard Deviation:", np.std(numbers))
print("Variance:", np.var(numbers))
print("Maximum:", np.max(numbers))
print("Minimum:", np.min(numbers))
print("Square:", np.square(numbers))
```

```python
from scipy import stats

numbers = [10, 20, 30, 40, 50]

print("Mean:", stats.tmean(numbers))
print("Mode:", stats.mode(numbers))
print("Standard Deviation:", stats.tstd(numbers))
print("Z-Scores:", stats.zscore(numbers))
```

```python
import pandas as pd

# Creating a dataset
data = {
    'Name': ['Amit', 'Riya', 'Rahul', 'Priya', 'Neha'],
    'Age': [20, 21, 19, 22, 20],
    'Marks': [85, 90, 75, 95, 80]
}

df = pd.DataFrame(data)

print("Complete Dataset:")
print(df)

print("\nFirst Five Rows:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

print("\nAverage Marks:", df['Marks'].mean())
print("\nStudents Scoring Above 80:")
print(df[df['Marks'] > 80])
```

```python
import matplotlib.pyplot as plt

names = ['Amit', 'Riya', 'Rahul', 'Priya', 'Neha']
marks = [85, 90, 75, 95, 80]

# Bar Chart
plt.bar(names, marks)
plt.title("Student Marks")
plt.xlabel("Student Names")
plt.ylabel("Marks")
plt.show()

# Line Chart
plt.plot(names, marks, marker='o')
plt.title("Student Performance")
plt.xlabel("Student Names")
plt.ylabel("Marks")
plt.show()

# Pie Chart
plt.pie(marks, labels=names, autopct='%1.1f%%')
plt.title("Marks Distribution")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'Name': ['Amit', 'Riya', 'Rahul', 'Priya', 'Neha'],
    'Marks': [85, 90, 75, 95, 80],
    'Age': [20, 21, 19, 22, 20]
}

df = pd.DataFrame(data)

# Bar Plot
sns.barplot(x='Name', y='Marks', data=df)
plt.title("Student Marks")
plt.show()

# Histogram
sns.histplot(data=df, x='Marks', bins=5, kde=True)
plt.title("Distribution of Marks")
plt.show()

# Box Plot
sns.boxplot(x='Marks', data=df)
plt.title("Marks Distribution")
plt.show()

# Scatter Plot
sns.scatterplot(x='Age', y='Marks', data=df)
plt.title("Age vs Marks")
plt.show()
```
